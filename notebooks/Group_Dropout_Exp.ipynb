```
import json
import subprocess
from pathlib import Path
import pandas as pd

from satx.engine import load_training_config


CONFIG_PATHS = {
    "MS_SPATIAL_DROP_BASELINE": resolve_project_path("configs/ms_spatial_dropout_baseline.yaml"),
    "MS_SPATIAL_DROP_P01": resolve_project_path("configs/ms_spatial_dropout_p10.yaml"),
    "MS_SPATIAL_DROP_P03": resolve_project_path("configs/ms_spatial_dropout_p30.yaml"),
    "MS_SPATIAL_DROP_P05": resolve_project_path("configs/ms_spatial_dropout_p50.yaml"),
    "MS_SPATIAL_DROP_P07": resolve_project_path("configs/ms_spatial_dropout_p70.yaml"),
    "MS_SPATIAL_DROP_P09": resolve_project_path("configs/ms_spatial_dropout_p90.yaml"),
    "MS_SPATIAL_DROP_P10": resolve_project_path("configs/ms_spatial_dropout_p100.yaml"),
}

git_hash = subprocess.check_output(
    ["git", "rev-parse", "HEAD"],
    text=True,
).strip()

summary_rows = []
epoch_rows = []

for run_id, config_path in CONFIG_PATHS.items():
    config = load_training_config(config_path)
    run_dir = config.run_dir()
    history_path = run_dir / "metrics_history.json"
    checkpoint_path = run_dir / "best.pt"

    history = json.loads(history_path.read_text(encoding="utf-8"))

    summary_rows.append(
        {
            "run_id": run_id,
            "protocol": "spatial",
            "modality": "ms",
            "group_dropout_p": config.group_dropout_p,
            "dropout_seed": config.dropout_seed,
            "training_seed": config.seed,
            "best_epoch": history["best_epoch"],
            "best_val_accuracy": history["best_val_accuracy"],
            "best_val_macro_f1": history["best_val_macro_f1"],
            "config_path": config_path,
            "checkpoint_path": str(checkpoint_path),
            "metrics_path": str(history_path),
            "git_hash": git_hash,
            "status": "complete",
        }
    )

    for epoch_record in history["epochs"]:
        counts = epoch_record["train"].get("mask_counts") or {}
        epoch_rows.append(
            {
                "run_id": run_id,
                "group_dropout_p": config.group_dropout_p,
                "epoch": epoch_record["epoch"],
                "train_loss": epoch_record["train"]["loss"],
                "train_accuracy": epoch_record["train"]["accuracy"],
                "val_loss": epoch_record["val"]["loss"],
                "val_accuracy": epoch_record["val"]["accuracy"],
                "val_macro_f1": epoch_record["val"]["macro_f1"],
                "mask_none": counts.get("NONE", 0),
                "mask_visible": counts.get("VISIBLE", 0),
                "mask_red_edge_nir": counts.get("RED_EDGE_NIR", 0),
                "mask_atmospheric": counts.get("ATMOSPHERIC", 0),
                "mask_swir": counts.get("SWIR", 0),
            }
        )

results_dir = resolve_project_path("results/drop_out_experiment")
results_dir.mkdir(exist_ok=True)

summary_df = pd.DataFrame(summary_rows)
epoch_df = pd.DataFrame(epoch_rows)

summary_df.to_csv(
    results_dir / "w3_dropout_validation_summary.csv",
    index=False,
)
epoch_df.to_csv(
    results_dir / "w3_dropout_epoch_history.csv",
    index=False,
)

summary_df
```

```
FORMAL_RUN_IDS = [
    "MS_SPATIAL_DROP_BASELINE",
    "MS_SPATIAL_DROP_P01",
    "MS_SPATIAL_DROP_P03",
    "MS_SPATIAL_DROP_P05",
    "MS_SPATIAL_DROP_P07",
    "MS_SPATIAL_DROP_P09",
    "MS_SPATIAL_DROP_P10",
]

baseline_row = (summary_df[summary_df["run_id"] == "MS_SPATIAL_DROP_BASELINE"].iloc[0])

dropout_candidates = (
    summary_df[
        summary_df["run_id"].isin(
            [
                "MS_SPATIAL_DROP_P01",
                "MS_SPATIAL_DROP_P03",
                "MS_SPATIAL_DROP_P05",
                "MS_SPATIAL_DROP_P07",
                "MS_SPATIAL_DROP_P09",
                "MS_SPATIAL_DROP_P10",
            ]
        )
    ]
    .sort_values(
        ["best_val_macro_f1", "group_dropout_p"],
        ascending=[False, True],
        kind="mergesort",
    )
)

selected_row = dropout_candidates.iloc[0]

display(
    dropout_candidates[
        [
            "run_id",
            "group_dropout_p",
            "best_epoch",
            "best_val_accuracy",
            "best_val_macro_f1",
            "checkpoint_path",
        ]
    ]
)

print("Selected dropout model:", selected_row["run_id"])
print("Selected p:", selected_row["group_dropout_p"])
print(
    "Selection validation macro-F1:",
    selected_row["best_val_macro_f1"],
)
```

```
import numpy as np
import torch

from sklearn.metrics import accuracy_score, f1_score

from satx.data import CLASS_NAMES
from satx.engine import TrainingConfig, resolve_device, set_seed
from satx.engine.data import build_test_dataloader
from satx.eval.fixed_group_mask import apply_fixed_group_mask
from satx.models import build_resnet50


def load_best_model_from_row(run_row):
    """Load and audit one best.pt selected by the validation summary."""
    device = resolve_device("auto")
    checkpoint_path = str(run_row["checkpoint_path"])

    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False,
    )

    checkpoint_config = TrainingConfig.from_dict(
        checkpoint["config"]
    )

    expected_best_epoch = int(run_row["best_epoch"])
    checkpoint_epoch = int(checkpoint["epoch"])

    if checkpoint_epoch != expected_best_epoch:
        raise RuntimeError(
            "Checkpoint epoch does not match validation summary: "
            f"checkpoint={checkpoint_epoch}, "
            f"summary={expected_best_epoch}"
        )

    checkpoint_val_macro_f1 = float(
        checkpoint["metrics"]["val"]["macro_f1"]
    )
    expected_val_macro_f1 = float(
        run_row["best_val_macro_f1"]
    )

    if not np.isclose(
        checkpoint_val_macro_f1,
        expected_val_macro_f1,
        rtol=0.0,
        atol=1e-12,
    ):
        raise RuntimeError(
            "Checkpoint validation macro-F1 does not match summary: "
            f"checkpoint={checkpoint_val_macro_f1}, "
            f"summary={expected_val_macro_f1}"
        )


    model = build_resnet50(
        num_classes=checkpoint_config.num_classes,
        in_channels=checkpoint_config.in_channels,
        pretrained=False,
        input_mode=checkpoint_config.model_input_mode,
    ).to(device)

    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    print(
        f"Loaded {run_row['run_id']}: "
        f"best.pt epoch={checkpoint_epoch}, "
        f"val macro-F1={checkpoint_val_macro_f1:.6f}"
    )

    return model, checkpoint_config, device


def evaluate_under_fixed_mask(
    model,
    test_loader,
    device,
    condition,
):
    valid_conditions = {
        "CLEAN",
        "VISIBLE",
        "RED_EDGE_NIR",
        "ATMOSPHERIC",
        "SWIR",
    }

    if condition not in valid_conditions:
        raise ValueError(
            f"Unknown condition {condition!r}. "
            f"Expected one of {sorted(valid_conditions)}."
        )

    model.eval()
    y_true = []
    y_pred = []

    with torch.inference_mode():
        for images, labels in test_loader:
            images = images.to(
                device,
                non_blocking=True,
            ).float()

            if condition != "CLEAN":
                images = apply_fixed_group_mask(
                    images,
                    condition,
                )

            logits = model(images)
            predictions = logits.argmax(dim=1)

            y_true.extend(labels.tolist())
            y_pred.extend(predictions.cpu().tolist())

    labels = list(range(len(CLASS_NAMES)))

    accuracy = accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average="macro",
        zero_division=0,
    )
    per_class_values = f1_score(
        y_true,
        y_pred,
        labels=labels,
        average=None,
        zero_division=0,
    )

    return {
        "condition": condition,
        "accuracy": float(accuracy),
        "macro_f1": float(macro_f1),
        "per_class_f1": {
            class_name: float(class_f1)
            for class_name, class_f1 in zip(
                CLASS_NAMES,
                per_class_values,
            )
        },
        "num_samples": len(y_true),
    }
```

```
baseline_model, baseline_config, baseline_device = (
    load_best_model_from_row(baseline_row)
)

selected_model, selected_config, selected_device = (
    load_best_model_from_row(selected_row)
)

assert baseline_device == selected_device
assert baseline_config.modality == selected_config.modality == "ms"
assert baseline_config.split_type == selected_config.split_type == "spatial"
assert baseline_config.normalization == selected_config.normalization == "zscore"

set_seed(baseline_config.seed)

test_loader = build_test_dataloader(baseline_config)

print("Spatial test samples:", len(test_loader.dataset))
print("Test batches:", len(test_loader))
```

```
import pandas as pd


TEST_CONDITIONS = [
    "CLEAN",
    "VISIBLE",
    "RED_EDGE_NIR",
    "ATMOSPHERIC",
    "SWIR",
]

MODEL_RUNS = [
    {
        "model_role": "baseline",
        "run_row": baseline_row,
        "model": baseline_model,
        "device": baseline_device,
    },
    {
        "model_role": "dropout_selected",
        "run_row": selected_row,
        "model": selected_model,
        "device": selected_device,
    },
]

test_summary_rows = []
test_class_rows = []

for model_run in MODEL_RUNS:
    model_role = model_run["model_role"]
    run_row = model_run["run_row"]
    model = model_run["model"]
    device = model_run["device"]

    for condition in TEST_CONDITIONS:
        print(
            f"Testing {run_row['run_id']} "
            f"under condition={condition}"
        )

        result = evaluate_under_fixed_mask(
            model=model,
            test_loader=test_loader,
            device=device,
            condition=condition,
        )

        test_summary_rows.append(
            {
                "model_role": model_role,
                "run_id": run_row["run_id"],
                "training_group_dropout_p": float(
                    run_row["group_dropout_p"]
                ),
                "selection_metric": (
                    "clean_spatial_validation_macro_f1"
                ),
                "selected_run_id": selected_row["run_id"],
                "selected_group_dropout_p": float(
                    selected_row["group_dropout_p"]
                ),
                "selected_val_macro_f1": float(
                    selected_row["best_val_macro_f1"]
                ),
                "test_condition": condition,
                "masked_group": (
                    "" if condition == "CLEAN" else condition
                ),
                "accuracy": result["accuracy"],
                "macro_f1": result["macro_f1"],
                "num_samples": result["num_samples"],
                "checkpoint_path": run_row["checkpoint_path"],
            }
        )

        for class_name, class_f1 in result[
            "per_class_f1"
        ].items():
            test_class_rows.append(
                {
                    "model_role": model_role,
                    "run_id": run_row["run_id"],
                    "training_group_dropout_p": float(
                        run_row["group_dropout_p"]
                    ),
                    "selected_run_id": selected_row["run_id"],
                    "selected_group_dropout_p": float(
                        selected_row["group_dropout_p"]
                    ),
                    "test_condition": condition,
                    "masked_group": (
                        "" if condition == "CLEAN" else condition
                    ),
                    "class_name": class_name,
                    "class_f1": class_f1,
                    "num_samples": result["num_samples"],
                    "checkpoint_path": run_row["checkpoint_path"],
                }
            )

print("All 10 test evaluations completed.")
```

```
masked_metrics_df = pd.DataFrame(test_summary_rows)
per_class_f1_df = pd.DataFrame(test_class_rows)

clean_overall = (
    masked_metrics_df[
        masked_metrics_df["test_condition"] == "CLEAN"
    ][
        ["run_id", "accuracy", "macro_f1"]
    ]
    .rename(
        columns={
            "accuracy": "clean_accuracy",
            "macro_f1": "clean_macro_f1",
        }
    )
)

masked_metrics_df = masked_metrics_df.merge(
    clean_overall,
    on="run_id",
    how="left",
    validate="many_to_one",
)

masked_metrics_df["accuracy_drop_from_clean"] = (
    masked_metrics_df["clean_accuracy"]
    - masked_metrics_df["accuracy"]
)

masked_metrics_df["macro_f1_drop_from_clean"] = (
    masked_metrics_df["clean_macro_f1"]
    - masked_metrics_df["macro_f1"]
)


baseline_by_condition = (
    masked_metrics_df[
        masked_metrics_df["model_role"] == "baseline"
    ][
        [
            "test_condition",
            "accuracy",
            "macro_f1",
            "accuracy_drop_from_clean",
            "macro_f1_drop_from_clean",
        ]
    ]
    .rename(
        columns={
            "accuracy": "baseline_accuracy",
            "macro_f1": "baseline_macro_f1",
            "accuracy_drop_from_clean": (
                "baseline_accuracy_drop_from_clean"
            ),
            "macro_f1_drop_from_clean": (
                "baseline_macro_f1_drop_from_clean"
            ),
        }
    )
)

masked_metrics_df = masked_metrics_df.merge(
    baseline_by_condition,
    on="test_condition",
    how="left",
    validate="many_to_one",
)

masked_metrics_df["accuracy_vs_baseline"] = (
    masked_metrics_df["accuracy"]
    - masked_metrics_df["baseline_accuracy"]
)

masked_metrics_df["macro_f1_vs_baseline"] = (
    masked_metrics_df["macro_f1"]
    - masked_metrics_df["baseline_macro_f1"]
)

# Positive value means dropout reduced the damage caused by masking.
masked_metrics_df["macro_f1_drop_reduction_vs_baseline"] = (
    masked_metrics_df["baseline_macro_f1_drop_from_clean"]
    - masked_metrics_df["macro_f1_drop_from_clean"]
)


clean_per_class = (
    per_class_f1_df[
        per_class_f1_df["test_condition"] == "CLEAN"
    ][
        ["run_id", "class_name", "class_f1"]
    ]
    .rename(columns={"class_f1": "clean_class_f1"})
)

per_class_f1_df = per_class_f1_df.merge(
    clean_per_class,
    on=["run_id", "class_name"],
    how="left",
    validate="many_to_one",
)

per_class_f1_df["class_f1_drop_from_clean"] = (
    per_class_f1_df["clean_class_f1"]
    - per_class_f1_df["class_f1"]
)

baseline_class_by_condition = (
    per_class_f1_df[
        per_class_f1_df["model_role"] == "baseline"
    ][
        ["test_condition", "class_name", "class_f1"]
    ]
    .rename(columns={"class_f1": "baseline_class_f1"})
)

per_class_f1_df = per_class_f1_df.merge(
    baseline_class_by_condition,
    on=["test_condition", "class_name"],
    how="left",
    validate="many_to_one",
)

per_class_f1_df["class_f1_vs_baseline"] = (
    per_class_f1_df["class_f1"]
    - per_class_f1_df["baseline_class_f1"]
)

# Save formal artifacts
masked_metrics_path = resolve_project_path(
    "results/drop_out_experiment/masked_group_metrics.csv"
)
per_class_path = resolve_project_path(
    "results/drop_out_experiment/masked_group_per_class_f1.csv"
)

masked_metrics_df.to_csv(
    masked_metrics_path,
    index=False,
)

per_class_f1_df.to_csv(
    per_class_path,
    index=False,
)

print("Saved:", masked_metrics_path)
print("Saved:", per_class_path)

display(
    masked_metrics_df[
        [
            "model_role",
            "training_group_dropout_p",
            "test_condition",
            "accuracy",
            "macro_f1",
            "macro_f1_drop_from_clean",
            "macro_f1_vs_baseline",
            "macro_f1_drop_reduction_vs_baseline",
        ]
    ]
)

display(
    masked_metrics_df.pivot(
        index="test_condition",
        columns="model_role",
        values="macro_f1",
    ).loc[TEST_CONDITIONS]
)
```

```
import pandas as pd

from satx.utils.paths import resolve_project_path


original_metrics_path = resolve_project_path(
    "results/drop_out_experiment/masked_group_metrics.csv"
)

raw_metrics_df = pd.read_csv(original_metrics_path)

assert len(raw_metrics_df) == 10
assert set(raw_metrics_df["model_role"]) == {
    "baseline",
    "dropout_selected",
}

TEST_CONDITION_ORDER = [
    "CLEAN",
    "VISIBLE",
    "RED_EDGE_NIR",
    "ATMOSPHERIC",
    "SWIR",
]

baseline_results = (
    raw_metrics_df[
        raw_metrics_df["model_role"] == "baseline"
    ]
    .set_index("test_condition")
    .loc[TEST_CONDITION_ORDER]
)

dropout_results = (
    raw_metrics_df[
        raw_metrics_df["model_role"] == "dropout_selected"
    ]
    .set_index("test_condition")
    .loc[TEST_CONDITION_ORDER]
)

comparison_numeric_df = pd.DataFrame(
    {
        "test_condition": TEST_CONDITION_ORDER,

        "baseline_accuracy": (
            baseline_results["accuracy"].to_numpy()
        ),
        "dropout_accuracy": (
            dropout_results["accuracy"].to_numpy()
        ),
        "accuracy_difference_pp": (
            (
                dropout_results["accuracy"].to_numpy()
                - baseline_results["accuracy"].to_numpy()
            )
            * 100
        ),

        "baseline_macro_f1": (
            baseline_results["macro_f1"].to_numpy()
        ),
        "dropout_macro_f1": (
            dropout_results["macro_f1"].to_numpy()
        ),
        "macro_f1_difference_pp": (
            (
                dropout_results["macro_f1"].to_numpy()
                - baseline_results["macro_f1"].to_numpy()
            )
            * 100
        ),

        "baseline_macro_f1_drop_pp": (
            baseline_results[
                "macro_f1_drop_from_clean"
            ].to_numpy()
            * 100
        ),
        "dropout_macro_f1_drop_pp": (
            dropout_results[
                "macro_f1_drop_from_clean"
            ].to_numpy()
            * 100
        ),
        "robustness_gain_pp": (
            (
                baseline_results[
                    "macro_f1_drop_from_clean"
                ].to_numpy()
                - dropout_results[
                    "macro_f1_drop_from_clean"
                ].to_numpy()
            )
            * 100
        ),
    }
)



comparison_display_df = comparison_numeric_df.copy()

score_columns = [
    "baseline_accuracy",
    "dropout_accuracy",
    "baseline_macro_f1",
    "dropout_macro_f1",
]

pp_columns = [
    "accuracy_difference_pp",
    "macro_f1_difference_pp",
    "baseline_macro_f1_drop_pp",
    "dropout_macro_f1_drop_pp",
    "robustness_gain_pp",
]

for column in score_columns:
    comparison_display_df[column] = (
        comparison_display_df[column]
        .map(lambda value: f"{value:.4f}")
    )

for column in pp_columns:
    comparison_display_df[column] = (
        comparison_display_df[column]
        .map(lambda value: f"{value:+.2f} pp")
    )

long_metrics_path = resolve_project_path(
    "results/drop_out_experiment/masked_group_metrics_long.csv"
)

raw_metrics_df.to_csv(
    long_metrics_path,
    index=False,
)


comparison_display_df.to_csv(
    original_metrics_path,
    index=False,
)


print("Saved compact comparison:", original_metrics_path)
print("Saved numeric long table:", long_metrics_path)

display(comparison_display_df)
```
